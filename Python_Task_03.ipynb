```python
# =====================================
# Python Practice - Q1
# =====================================

# Question:
# Given the following list:
#
# numbers = [12, 7, 20, 15, 8, 25, 10, 30]
#
# Calculate the average of the numbers that are greater than 10.
# Do not use sum().
```

```python
numbers = [12, 7, 20, 15, 8, 25, 10, 30]

total = 0
count = 0

for i in numbers:
    if i > 10:
        total += i
        count += 1

print(total / count)
```

```python
# =====================================
# Python Practice - Q2
# =====================================

# Question:
# Given the following dictionary:
#
# students = {
#     "Ahmed": 85,
#     "Sara": 92,
#     "Ali": 74,
#     "Mona": 88,
#     "Omar": 65
# }
#
# Find the student with the highest grade.
# Print the student's name and grade.
# Do not use max().
```

```python
students = {
    "Ahmed": 85,
    "Sara": 92,
    "Ali": 74,
    "Mona": 88,
    "Omar": 65
}

highest = 0
name = ""

for i in students:
    if students[i] > highest:
        highest = students[i]
        name = i

print(name, highest)
```

```python
# =====================================
# Python Practice - Q3
# =====================================

# Question:
# Given the following tuple:
#
# numbers = (10, 15, 20, 25, 30, 35, 40)
#
# Create a new tuple containing only the numbers
# that are divisible by 5 and greater than 20.
```

```python
numbers = (10, 15, 20, 25, 30, 35, 40)

result = ()

for i in numbers:
    if i > 20 and i % 5 == 0:
        result += (i,)

print(result)
```

```python
# =====================================
# Python Practice - Q5
# =====================================

# Question:
# Given the following dictionary:
#
# products = {
#     "Laptop": 25000,
#     "Phone": 15000,
#     "Tablet": 12000,
#     "Monitor": 8000,
#     "Keyboard": 1500
# }
#
# Create a new dictionary containing only products
# with prices between 5000 and 20000 inclusive.
```

```python
products = {
    "Laptop": 25000,
    "Phone": 15000,
    "Tablet": 12000,
    "Monitor": 8000,
    "Keyboard": 1500
}

result = {}

for i in products:
    if 5000 <= products[i] <= 20000:
        result[i] = products[i]

print(result)
```

```python
# =====================================
# Python Practice - Q8
# =====================================

# Question:
# Given the following list:
#
# numbers = [5, 12, 8, 21, 30, 17, 10, 25]
#
# Find the smallest and largest numbers
# without using min() or max().
```

```python
numbers = [5, 12, 8, 21, 30, 17, 10, 25]

small = numbers[0]
big = numbers[0]

for i in numbers:
    if i < small:
        small = i
    if i > big:
        big = i

print(small)
print(big)
```

```python
# =====================================
# Python Practice - Q10
# =====================================

# Question:
# Given the following list:
#
# numbers = [10, 20, 10, 30, 40, 20, 50, 30, 60]
#
# Create a new list containing each number only once.
#
# The original order must be preserved.
#
# Expected output:
# [10, 20, 30, 40, 50, 60]
```

```python
# =====================================
# Python Practice - Q11
# =====================================

# Question:
# Given the following tuple:
#
# students = (
#     ("Ahmed", 85),
#     ("Sara", 92),
#     ("Ali", 78),
#     ("Mona", 95)
# )
#
# Calculate the average grade of all students.
# Then print the names of students whose grade
# is greater than the average.
```

```python
students = (
    ("Ahmed", 85),
    ("Sara", 92),
    ("Ali", 78),
    ("Mona", 95)
)

total = 0

for name, grade in students:
    total += grade

avg = total / len(students)

print(avg)

for name, grade in students:
    if grade > avg:
        print(name)
```

```python
# =====================================
# Python Practice - Q12
# =====================================

# Question:
# Write a program that asks the user to enter numbers
# repeatedly.
#
# Stop when the user enters 0.
#
# After stopping, print:
# 1. The total number of entered values.
# 2. The sum of the values.
# 3. The average of the values.
#
# Do not include 0 in the calculations.
```

```python
total = 0
count = 0

while True:
    num = int(input("Enter number: "))

    if num == 0:
        break

    total += num
    count += 1

print(count)
print(total)

if count > 0:
    print(total / count)
else:
    print(0)
```

```python
# =====================================
# Python Practice - Q13
# =====================================

# Question:
# Given the following dictionary:
#
# inventory = {
#     "Laptop": 5,
#     "Phone": 0,
#     "Tablet": 8,
#     "Monitor": 0,
#     "Keyboard": 12
# }
#
# Find all products that are out of stock.
# Then count how many products are out of stock.
```

```python
inventory = {
    "Laptop": 5,
    "Phone": 0,
    "Tablet": 8,
    "Monitor": 0,
    "Keyboard": 12
}

count = 0

for i in inventory:
    if inventory[i] == 0:
        print(i)
        count += 1

print(count)
```

```python
# =====================================
# Python Practice - Q14
# =====================================

# Question:
# Given the following string:
#
# text = "Python is easy and Python is powerful"
#
# Count how many times each word appears.
#
# Expected result:
# {
#     "Python": 2,
#     "is": 2,
#     "easy": 1,
#     "and": 1,
#     "powerful": 1
# }
```

```python
text = "Python is easy and Python is powerful"

words = text.split()
result = {}

for i in words:
    if i in result:
        result[i] += 1
    else:
        result[i] = 1

print(result)
```

```python
# =====================================
# Python Practice - Q15
# =====================================

# Question:
# Given the following list:
#
# numbers = [4, 7, 2, 9, 10, 15, 20, 3]
#
# Create two separate lists:
#
# 1. Numbers divisible by 2.
# 2. Numbers divisible by 3.
#
# If a number is divisible by both 2 and 3,
# it should appear in both lists.
#
# Print both lists.
```

```python
numbers = [4, 7, 2, 9, 10, 15, 20, 3]

a = []
b = []

for i in numbers:
    if i % 2 == 0:
        a.append(i)

    if i % 3 == 0:
        b.append(i)

print(a)
print(b)
```
